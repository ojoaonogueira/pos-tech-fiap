```python
from sklearn.preprocessing import LabelEncoder
```

```python
df.head()
```

```python
colunas = ['gender','workex','specialisation','status']
```

```python
label_encoder = LabelEncoder()
```

```python
for col in colunas:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
df.head()
```

```python
dummy_hsc_s = pd.get_dummies(df['hsc_s'],prefix="dummy")
```

```python
dummy_degree_t = pd.get_dummies(df['degree_t'],prefix="dummy")
```

```python
df_dummy = pd.concat([df, dummy_hsc_s, dummy_degree_t], axis=1)
```

```python
df_dummy.head()
```

```python
df_dummy.drop(['hsc_s', 'degree_t', 'salary'], axis=1, inplace=True)
```

```python
df_dummy.head()
```

```python
correlation_matrix = df_dummy.select_dtypes(include=['number', 'bool']).corr().round(2)
fig, ax = plt.subplots(figsize=(12,8))
sns.heatmap(data=correlation_matrix, annot=True, linewidths=5, ax=ax, cmap='coolwarm')
```

```python
x = df_dummy[['ssc_p','hsc_p', 'degree_p', 'workex', 'mba_p']]
y = df_dummy['status']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=7, stratify=y)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
scaler = StandardScaler()
scaler.fit(x_train)
x_train_escalnodado = scaler.transform(x_train)
x_test_escalnodado = scaler.transform(x_test)
```

```python
x_train_escalnodado
```

```python
erros = []
for i in range(1,10):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train_escalnodado, y_train)
    pred_i = knn.predict(x_test_escalnodado)
    erros.append(np.mean(pred_i != y_test))
```

```python
plt.figure(figsize=(12,6))
plt.plot(range(1,10), erros, color='red', linestyle='dashed', marker='o', markerfacecolor='blue', markersize=10)
plt.title('Erro medio para K')
plt.ylabel('Erro medio')
plt.xlabel('valor de k')
plt.show()
```

```python
modelo_classificador = KNeighborsClassifier(n_neighbors=5)

modelo_classificador.fit(x_train_escalnodado, y_train)
y_predito = modelo_classificador.predict(x_test_escalnodado)
```

```python
y_predito
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test, y_predito))
```

```python
from sklearn.svm import LinearSVC
```

```python
from sklearn.pipeline import Pipeline
```

```python
svm = Pipeline(
    [
        ("linear_svc", LinearSVC(C=1))
    ]
)
```

```python
svm.fit(x_train_escalnodado, y_train)
```

```python
y_predito_svm = svm.predict(x_test_escalnodado)
```

```python
print(accuracy_score(y_test, y_predito_svm))
```
